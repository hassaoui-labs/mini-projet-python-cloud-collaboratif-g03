**Fonction moyenne**  
- Permet de calculer la moyenne d'une liste donnée en argument.  
- La fonction provoque une erreur ZeroDivisionError si la liste est vide

```python
def moyenne(liste):
  return sum(liste) / len(liste) # La moyenne est égale à la somme des éléments de la liste sur le nombre d'éléments

print(moyenne([])) # Appel de la fonction moyenne avec une liste vide donnée en argument
```

**Version débogage avec debugpy**

```python
import debugpy

debugpy.listen(("0.0.0.0", 5678))
debugpy.wait_for_client()  # bloque ici

def moyenne(liste):
    debugpy.breakpoint()
    return sum(liste) / len(liste)

print(moyenne([]))
```
