**Fonction carre**
- Permet de calculer le carré d'un nombre donné en argument
- La fonction provoque une erreur TypeError lorsque la valeur donnée comme argument n'est pas de type numérique (int, float, ou complexe)

```python
def carre(x):
    if not isinstance(x, (int, float, complex)):
        raise TypeError("x doit être un nombre (int, float ou complex)")
    return x**2

try:
    x = float(input("Entrez un nombre : "))
    print(carre(x))
except ValueError:
    print("Erreur: veuillez entrer un nombre valide.")
```

Reproduction volontaire du bug

```python
# Cas volontairement incorrect pour illustrer le debugging
try:
    x = input("Entrez un nombre : ")  # pas de conversion volontaire
    print(carre(x))
except Exception as e:
    print("Erreur détectée :", e)
```

**Debugging**

```python
import pdb

def carre_debug(x_dbg):
    pdb.set_trace()  # point d’arrêt
    if not isinstance(x_dbg, (int, float, complex)):
        raise TypeError("x doit être un nombre (int, float ou complex)")
    return x_dbg**2

try:
    x_dbg = input("Entrez un nombre (debug) : ")  # volontairement non converti
    print("Résultat :", carre_debug(x_dbg))
except Exception as e:
    print("Erreur détectée (debug) :", e)
```
