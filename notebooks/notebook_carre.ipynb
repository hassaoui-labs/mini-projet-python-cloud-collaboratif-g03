**Fonction carre**
- Permet de calculer le carré d'un nombre donné en argument
- La fonction provoque une erreur TypeError lorsque la valeur donnée comme argument n'est pas de type numérique (int, float, ou complexe)

```python
def carre(x): # Déclarer la fonction carre
  return x**2

x = input("Entrez un nombre : ") # Lire la valeur de x
print(carre(x)) # Appel de la fonction carre avec x comme argument
```

**Version débogage avec debugpy**

```python
import debugpy

debugpy.listen(("0.0.0.0", 5678))
debugpy.wait_for_client()  # bloque ici tant qu’aucun debugger n’est attaché

def carre(x):  # Déclarer la fonction carre
    debugpy.breakpoint()   # point d’arrêt dans la fonction
    return x**2

x = input("Entrez un nombre : ")  # Lire la valeur de x
print(carre(x))                  # Appel de la fonction carre avec x comme argument
```
