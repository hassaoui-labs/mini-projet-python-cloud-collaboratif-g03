**Fonction lire_fichier**
- Permet de lire le contenu d'un fichier.
- La fonction provoque une erreur FileNotFoundError si le fichier n'existe pas.

```python
def lire_fichier(fichier):
    try:
        with open(fichier, "r") as f:
            return f.readlines()
    except FileNotFoundError:
        return [f"Erreur: le fichier '{fichier}' n'existe pas.\n"]

print(lire_fichier("fichier.txt"))
```

Création du fichier (correction)

```python
# Création du fichier pour corriger l'erreur
with open("fichier.txt", "w") as f:
    f.write("Ligne 1\nLigne 2\nLigne 3\n")

print("Fichier créé avec succès.")
```

Vérification après correction

```python
# Relecture après correction
print(lire_fichier("fichier.txt"))
```

```python
import pdb

def lire_fichier_debug(fichier):
    pdb.set_trace()
    try:
        with open(fichier, "r") as f:
            return f.readlines()
    except FileNotFoundError:
        return [f"Erreur : le fichier '{fichier}' n'existe pas.\n"]

print(lire_fichier_debug("fichier.txt"))
```
