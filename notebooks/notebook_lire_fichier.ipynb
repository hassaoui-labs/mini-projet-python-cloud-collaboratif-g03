**Fonction lire_fichier**
- Permet de lire le contenu d'un fichier.
- La fonction provoque une erreur FileNotFoundError si le fichier n'existe pas.

```python
def lire_fichier(fichier):
  with open(fichier, 'r') as f: # Ouvrir le fichier avec mode lecture
    lignes = f.readlines()
  return lignes

print(lire_fichier('fichier.txt')) # Appel de la fonction lire_fichier
```
