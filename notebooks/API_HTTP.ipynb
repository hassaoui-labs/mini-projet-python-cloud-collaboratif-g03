**1) Fonction fetch_posts**

*   Permet de récupérer des données (posts) depuis une API.

*   La fonction provoque une erreur HTTPError / 404 si l’URL est fausse



```python
import requests

def fetch_posts(user_id):
    try:
        url = "https://jsonplaceholder.typicode.com/postsS"
        r = requests.get(url, params={"userId": user_id})

        r.raise_for_status()

        return r.json()

    except requests.exceptions.HTTPError:
        return [f"Erreur: requête HTTP échouée (status={r.status_code}).\n"]

print(fetch_posts(1))
```

```python
# Correction : URL correcte
def fetch_posts_fixed(user_id):
    try:
        url = "https://jsonplaceholder.typicode.com/posts"
        r = requests.get(url, params={"userId": user_id})
        r.raise_for_status()
        return r.json()

    except requests.exceptions.HTTPError:
        return [f"Erreur: requête HTTP échouée (status={r.status_code}).\n"]

print("Correction appliquée avec succès.")
```

2) **Debugging** avec pdb **texte en gras**

```python
import pdb
import requests

def fetch_posts_debug(user_id):
    pdb.set_trace()
    r = None
    try:
        url = "https://jsonplaceholder.typicode.com/posts"
        r = requests.get(url, params={"userId": user_id}, timeout=10)
        r.raise_for_status()
        return r.json()

    except requests.exceptions.HTTPError:
        return [f"Erreur: requête HTTP échouée (status={r.status_code}).\n"]
    except requests.exceptions.Timeout:
        return ["Erreur: Timeout (API trop lente).\n"]
    except requests.exceptions.ConnectionError:
        return ["Erreur: Problème de connexion réseau.\n"]

print(fetch_posts_debug(1)[:1])

```
